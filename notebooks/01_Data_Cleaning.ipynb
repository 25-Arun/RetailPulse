```python
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully.")
```

```python
file_path = "../data/online_retail_II.xlsx"

# Load Sheet 1 and save as CSV
df_sheet1 = pd.read_excel(file_path, sheet_name=0)
df_sheet1.to_csv("../data/sheet1.csv", index=False)

# Load Sheet 2 and save as CSV
df_sheet2 = pd.read_excel(file_path, sheet_name=1)
df_sheet2.to_csv("../data/sheet2.csv", index=False)

df1 = pd.read_csv("../data/sheet1.csv")
df2 = pd.read_csv("../data/sheet2.csv")
```

```python
df = pd.concat([df1, df2], ignore_index=True)
```

```python
display(df.head())

print("\n--- DATA TYPES & MISSING VALUES ---")
df.info()

print("\n--- SUMMARY STATISTICS ---")
display(df.describe(include='all'))

print(f"\nMissing Values:\n{df.isnull().sum()}")
print(f"\nDuplicate Rows: {df.duplicated().sum()}")
```

```python

df.rename(columns={
    "Invoice": "InvoiceNo",
    "Customer ID": "CustomerID",
    "Price": "UnitPrice"
}, inplace=True)


df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df = df[
    (~df["InvoiceNo"].astype(str).str.startswith("C")) &  
    (df["Quantity"] > 0) &                               
    (df["UnitPrice"] > 0)                                
].dropna(subset=["Description", "CustomerID"]).drop_duplicates()  


df["TotalSales"] = df["Quantity"] * df["UnitPrice"]
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["MonthName"] = df["InvoiceDate"].dt.month_name()
df["Quarter"] = df["InvoiceDate"].dt.quarter
df["Week"] = df["InvoiceDate"].dt.isocalendar().week
df["Day"] = df["InvoiceDate"].dt.day
df["DayName"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour
df["Weekend"] = df["DayName"].isin(["Saturday", "Sunday"])

print(f"Cleaning & Feature Engineering Complete. Final Shape: {df.shape}")
```
